```python
import csv,sqlite3
import prettytable
prettytable.DEFAULT = 'DEFAULT'

con = sqlite3.connect('my_data.db')
cur = con.cursor()
```

```python
%sql sqlite:///my_data.db
```

```python
import pandas as pd
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv")
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False,method="multi")
```

```python
%sql drop table if exists SPACEXTBLE
```

```python
%sql create table SPACEXTBLE as select * from SPACEXTBL where Date is not null
```

```python
%sql SELECT DISTINCT(Launch_Site) FROM SPACEXTBLE
```

```python
%sql SELECT * FROM SPACEXTBLE WHERE launch_site LIKE 'CCA%' LIMIT 5
```

```python
%sql select * from SPACEXTBLE LIMIT 5
```

```python
%sql select SUM(PAYLOAD_MASS__KG_) from SPACEXTBLE where Customer = 'NASA (CRS)'
```

```python
%sql select AVG(PAYLOAD_MASS__KG_) FROM SPACEXTBLE where Booster_Version like 'F9 v1.1%'
```

```python
%sql select min(Date) from SPACEXTBLE where landing_outcome = 'Success (ground pad)'
```

```python
%sql select Booster_Version from SPACEXTBLE where landing_outcome = 'Success (drone ship)' and PAYLOAD_MASS__KG_ between 4001 and 5999
```

```python
%sql select Mission_Outcome, count(Mission_Outcome) as Outcome from SPACEXTBLE group by Mission_Outcome
```

```python
%sql select Booster_Version from SPACEXTBLE where PAYLOAD_MASS__KG_ = (select max(PAYLOAD_MASS__KG_) from SPACEXTBLE)
```

```python
%sql select Date,Booster_Version,Launch_Site,Landing_Outcome from SPACEXTBLE where Landing_Outcome = 'Faure (drone ship)' and year(Date) =2015
```

```python
%sql select Landing_Outcome,COUNT(*) as Count_Launches from SPACEXTBLE where Date between '2010-06-04' and '2017-03-20' group by Landing_Outcome order by Count_Launches DESC
```
